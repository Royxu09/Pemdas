##**Kasus Soal Kasus Analisis Perulangan Gacha TCG**

Komunitas Pokémon TCG kamu punya bot di Discord. Anggota komunitas pengen pamer hasil gacha mereka kemarin sore. Mereka bakal masukin 5 nama kartu yang mereka dapet ke bot itu.


#**PROBLEM STATEMENT**
Bagaimana cara memfasilitasi anggota komunitas Pokémon TCG agar dapat memamerkan (pamer) hasil gacha kartu mereka secara praktis, visual, dan interaktif melalui Discord?

```python
def cek_rarity_kartu(nama_kartu):
  jenis_kartu = {
      "Pikachu": "Common",
      "Bulbasaur": "Common",
      "Charmander": "Uncommon",
      "Gengar": "Rare",
      "Mewtwo": "SR",
      "Rayquaza": "SAR/UR",
      "Golden": "God Pack"
  }
  rarity = jenis_kartu.get(nama_kartu)
  if rarity == "Common":
    return f"[{rarity}] Yahh Ampas banget Common Doang"
  elif rarity == "Uncommon":
    return f"[{rarity}] Aduh masa Uncommon doang"
  elif rarity == "Rare":
    return f"[{rarity}] Ah lumayanlah"
  elif rarity == "AR":
    return f"[{rarity}] Wah AR nih bos"
  elif rarity == "SR":
    return f"[{rarity}] WOW WANGYY SR NIH"
  elif rarity == "SAR/UR":
    return f"[{rarity}] WAH GACOR BANGET PACK INI"
  elif rarity == "God Pack":
    return f"[{rarity}] WAHHHHH DI LUAR NALAR SIH INI GOD PACK GOKILLL BANGETTTT"
  else:
    return "Error"

print("== BUKA PACK TCG KAMU (Isi 5 Kartu)")
keranjang_kartu = []
for i in range(5):
  nama_kartu = input (f"Masukkan nama kartu ke - {i + 1}: ")
  keranjang_kartu.append(nama_kartu)

print("-" * 30)
print("Hasil Buka Pack Kamu: ")
print("-" * 30)

for i in range(len(keranjang_kartu)):
  kartu = keranjang_kartu[i]
  hasil = cek_rarity_kartu(kartu)
  print(f"Kartu {i+1}: {hasil}")
```
